### **NUMBER NORMALIZER**

```python
import re
import inflect

converter = inflect.engine()

currency_map = {
    "$": "dollars,cents",
    "€": "euros,cents",
    "£": "pounds,pence",
    "¥": "yen,sen",
    "₹": "rupees,paise",
    "¢": "cent"
}

def convert_currency_with_suffix(match):
    symbol, amount_str, suffix = match.group(1), match.group(2), match.group(3).lower()
    try:
        amount = float(amount_str)
        amount_words = converter.number_to_words(amount, andword="and")
    except ValueError:
        return match.group(0)
    units = currency_map.get(symbol, "").split(",")
    major_unit = units[0]
    suffix_word = "billion" if suffix == "bn" else "million"
    return f"{amount_words} {suffix_word} {major_unit if amount == 1 else converter.plural(major_unit)}"

def convert_suffix(match):
    num = match.group(1)
    suffix = match.group(2).lower()


    surrounding = match.string[match.end():match.end()+10].lower()
    if suffix == "m" and any(x in surrounding for x in ["meter", "metres", "m ", "m-long", "m wide", "m high" , "100m title", "100m Title", "100m race"]):
        return match.group(0)

    try:
        num = float(num)
        word_number = converter.number_to_words(num, andword="and")
        suffix_word = "billion" if suffix == "bn" else "million"
        return f"{word_number} {suffix_word}"
    except:
        return match.group(0)


def convert_percent(match):
    try:
        num = float(match.group(1))
        word = converter.number_to_words(num, andword="and")
        return f"{word} percent"
    except:
        return match.group(0)

def convert_plain_numbers(match):
    num = match.group(0)
    try:
        if '.' in num:
            whole, frac = map(int, num.split('.'))
            return f"{converter.number_to_words(whole)} point {converter.number_to_words(frac)}"
        else:
            return converter.number_to_words(int(num))
    except:
        return num

def convert_currency(match):
    symbol, amount_str = match.group(1), match.group(2)
    try:
        amount = float(amount_str)
    except:
        return match.group(0)
    units = currency_map.get(symbol, "").split(",")
    major_unit = units[0]
    minor_unit = units[1] if len(units) > 1 else None
    whole = int(amount)
    fraction = int(round((amount - whole) * 100))
    major_word = converter.number_to_words(whole, andword="and")
    result = f"{major_word} {major_unit if whole == 1 else converter.plural(major_unit)}"
    if minor_unit and fraction > 0:
        minor_word = converter.number_to_words(fraction, andword="and")
        result += f" and {minor_word} {minor_unit if fraction == 1 else converter.plural(minor_unit)}"
    return result
```

```python
def convert_all_numbers(text):
    text = re.sub(r'([€£¥₹$¢])\s*([0-9]+(?:\.[0-9]+)?)\s*(bn|m)\b', convert_currency_with_suffix, text, flags=re.IGNORECASE)
    text = re.sub(r'\b([0-9]+(?:\.[0-9]+)?)(bn|m)\b', convert_suffix, text, flags=re.IGNORECASE)
    text = re.sub(r'\b([0-9]+(?:\.[0-9]+)?)\s*%', convert_percent, text)
    text = re.sub(r'([€£¥₹$¢])\s*([0-9]+(?:\.[0-9]+)?)', convert_currency, text)
    text = re.sub(r'\b[0-9]+(?:\.[0-9]+)?\b', convert_plain_numbers, text)
    return text
```

### ***TEXT CLEANING***

Text cleaning - cleaning of text by removing special characters and replacing URL's and Numbers for better learning.

```python
def clean_text(text):
    text = re.sub(r'\W', ' ', text)      # Remove special characters
    text = re.sub(r'\s+', ' ', text)     # Remove extra spaces
    text = re.sub(r'http\S+|www.\S+', '<URL>', text) #replace URL's with the word URL for cleaner text
    text = re.sub(r'http\S+|xxxmobilemovieclub.\S+', '<URL>', text) #remove other URL's
    #text = re.sub(r'\d+', ' <NUM>', text) #remove numbers
    text = re.sub(r'[^\x00-\x7F]', ' ', text) #removes non-ASCII characters
    text = text.strip()  # Remove leading/trailing spaces
    text = re.sub(r'[^\w\s]', '', text) #punctuation
    return text
```

```python
def get_core_vocab(counter, coverage=0.9):
    total = sum(counter.values())
    cumulative = 0
    core_vocab = []
    for word, count in counter.most_common():
        cumulative += count
        core_vocab.append(word)
        if cumulative / total >= coverage:
            break
    return core_vocab
```

```python
from gensim.models import Word2Vec
from gensim.models.callbacks import CallbackAny2Vec

class LossLogger(CallbackAny2Vec):
    def __init__(self):
        self.epoch = 0
        self.loss_previous_step = 0
        self.losses = []

    def on_epoch_end(self, model):
        loss = model.get_latest_training_loss()
        loss_diff = loss - self.loss_previous_step
        self.losses.append(loss_diff)
        print(f"Epoch {self.epoch} | Loss: {loss_diff}")
        self.loss_previous_step = loss
        self.epoch += 1
```

```python
loss_logger = LossLogger()
w2v_model = Word2Vec(
    sentences=bbcdf['general'],# Tokenized corpus: List of lists
    vector_size=300,           # Dimensionality of word embeddings
    window=5,                  # Context window size (left and right)
    min_count=5,               # Ignores words with total freq < 5
    workers=4,                 # Number of CPU threads to use
    sg=1,                      # 1 = Skip-gram; 0 = CBOW
    epochs=60,                 # Number of training iterations (default=5)
    alpha=0.025,               # Initial learning rate (use with epochs)
    min_alpha=0.0001,          # Final learning rate for training (linear decay)
    negative=10,               # Number of negative samples per positive example
    ns_exponent=0.75,          # Controls the distribution of negative sampling
    sample=1e-4,               # Downsampling threshold for frequent words
    hs=0,                      # 0 = use negative sampling, 1 = use hierarchical softmax
    seed=42,                   # Random seed for reproducibility
    max_final_vocab=None,      # Limits the final vocabulary size
    sorted_vocab=1,            # 1 = sort vocab by descending freq (default)
    compute_loss=True,         # Set True to track training loss
    callbacks=[loss_logger],   # List of callbacks (e.g., monitor training progress)
)

w2v_model.save("custom_word2vec.model")

plt.plot(loss_logger.losses)
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```

### GLove Model

```python


import numpy as np

# Load GloVe 300D vectors into a dictionary, can try different glove models with more parameters.
def load_glove_embeddings(glove_file_path):
    embeddings_index = {}
    with open(glove_file_path, encoding="utf8") as f:
        for line in f:
            values = line.split()
            word = values[0]
            vector = np.asarray(values[1:], dtype="float32")
            embeddings_index[word] = vector
    return embeddings_index

glove_path = "/content/glove.6B.300d.txt"
glove_embeddings = load_glove_embeddings(glove_path)
print("Loaded GloVe vectors:", len(glove_embeddings))
```

```python
def get_word_embeddings(tokens, embeddings_index, dim=300):
    word_vectors = []
    for word in tokens:
        if word in embeddings_index:
            word_vectors.append(embeddings_index[word])
        else:
            word_vectors.append(np.zeros(dim))  # Handle unknown words by returning zeros
    return word_vectors

bbcdf['glove_vector'] = bbcdf['general'].apply(lambda x: get_word_embeddings(x, glove_embeddings, 300))
```

```python
import pandas as pd

def extract_word_level_glove_vectors(df, token_col='general', vec_col='glove_vector'):
    word_vec_dict = {}
    for _, row in df.iterrows():
        tokens = row[token_col]
        vectors = row[vec_col]
        for token, vec in zip(tokens, vectors):
            if token not in word_vec_dict:
                word_vec_dict[token] = vec
    word_df = pd.DataFrame(word_vec_dict).T
    word_df.index.name = 'word'
    word_df.reset_index(inplace=True)
    return word_df


df_glove_vectors = extract_word_level_glove_vectors(bbcdf) #creating a new df with word level vectors for the entire vocab
```

```python
def encoding_layer(text_series, vocab_size, sequence_length):
    processed_text_series = text_series.apply(lambda x: ' '.join(x) if isinstance(x, list) else '')
    vectorizer = TextVectorization(
        max_tokens=vocab_size,
        output_mode="int",
        output_sequence_length=sequence_length
    )
    vectorizer.adapt(processed_text_series)
    encoded_texts = np.array([vectorizer(tf.constant([text])).numpy()[0] for text in processed_text_series])
    print(f"Encoded Shape: {encoded_texts.shape}")
    return vectorizer, encoded_texts
```
